## For example, when applying stemming to the sentence, "The fishers are fishing in the fished waters," a stemming algorithm strips the suffixes to yield:"The fisher are fish in the fish water"

```python
sen = "The fishers are fishing in the fished waters" 
print('before stemming :', sen)
```

```python
from nltk.tokenize import word_tokenize
wt = word_tokenize
wt
```

```python
tokens = wt(sen)
tokens
```

```python
res = ''
for i in tokens :
    res += ps.stem(i)
    res += ' '

print('after stemming :', res)
```

## Text Processing

```python
from nltk.corpus import stopwords
en_stopwords = set(stopwords.words('english'))
print(en_stopwords)
```

```python
ben_stopwords = set(stopwords.words('bengali'))
print(ben_stopwords)
```

```python
import string
punc = string.punctuation
punc
```

```python
len(punc)
```

```python
def preprocess_text(text) :
    remove_punc = [char for char in text if char not in punc]
    clean_words = ''.join(remove_punc)
    split_words = clean_words.split()
    text = [word for word in split_words if word.lower() not in en_stopwords]
    return text
```

```python
df
```

```python
df.columns
```

```python
tmp = df['class']
df.drop(columns = ['class'], inplace = True)
df['cleaned_text'] = df.text.apply(preprocess_text)
df['class'] = tmp
```

```python
df
```

```python
lemmatizer = WordNetLemmatizer()
lemmatizer
```

```python
def lemmatize_text(text_list):
    clean_text = []
    for word in text_list :
        clean_text.append(lemmatizer.lemmatize(word))

    return ' '.join(clean_text)


tmp = df['class']
df.drop(columns = ['class'], inplace = True)
df['lemmatize_text'] = df.cleaned_text.apply(lemmatize_text)
df['class'] = tmp
df
```
