# Stemming in NLP

```python
from nltk.tokenize import word_tokenize, sent_tokenize, PunktSentenceTokenizer
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
ps
```

```python
sen = "The server crashed due to a memory leak in the database."
sen
```

```python
word_tokens = word_tokenize(sen)
word_tokens
```

```python
for word in word_tokens :
    print(word, " --- ", ps.stem(word))
```

## Text Processing¶

```python
import string 
punc = string.punctuation
punc
```

```python
from nltk.corpus import stopwords
```

```python
en_stopwords = set(stopwords.words('english'))
en_stopwords
```

```python
ben_stopwords = set(stopwords.words('bengali'))
ben_stopwords
```

```python
df.columns
```

```python
def preprocess_text(text) :
    remove_punc = [char for char in text if char not in punc]
    clean_words = ''.join(remove_punc)
    split_words = clean_words.split()
    text = [word for word in split_words if word.lower() not in en_stopwords]
    return text
```

```python
df.text = df.text.apply(preprocess_text)
df
```

```python
def lemmatized_text(text) :
    clean_text = []
    for word in text :
        clean_text.append(wnl.lemmatize(word))

    return ' '.join(clean_text)
```

```python
df.text = df.text.apply(lemmatized_text)
df
```
